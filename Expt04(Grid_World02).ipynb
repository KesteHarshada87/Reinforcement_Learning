```python
import numpy as np

# Grid World size
rows, cols = 4, 4

# Terminal states
terminal_states = [(0, 0), (3, 3)]

# Actions: Up, Down, Left, Right
actions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
action_symbols = ['↑', '↓', '←', '→']

# Discount factor
gamma = 0.9

# Initialize policy randomly
policy = np.random.choice(4, size=(rows, cols))

# Value function
V = np.zeros((rows, cols))


# Function to perform one step
def get_next_state(state, action):
    i, j = state
    di, dj = actions[action]

    ni = i + di
    nj = j + dj

    # If outside grid, stay in same state
    if ni < 0 or ni >= rows or nj < 0 or nj >= cols:
        ni, nj = i, j

    return ni, nj


# Policy Iteration
while True:

    # -------------------------
    # 1. Policy Evaluation
    # -------------------------
    while True:
        delta = 0

        for i in range(rows):
            for j in range(cols):

                if (i, j) in terminal_states:
                    continue

                old_value = V[i, j]

                action = policy[i, j]

                ni, nj = get_next_state((i, j), action)

                reward = -1

                # Reaching goal
                if (ni, nj) == (3, 3):
                    reward = 0

                V[i, j] = reward + gamma * V[ni, nj]

                delta = max(delta, abs(old_value - V[i, j]))

        if delta < 0.001:
            break

    # -------------------------
    # 2. Policy Improvement
    # -------------------------
    policy_stable = True

    for i in range(rows):
        for j in range(cols):

            if (i, j) in terminal_states:
                continue

            old_action = policy[i, j]

            action_values = []

            for action in range(4):
                ni, nj = get_next_state((i, j), action)

                reward = -1

                if (ni, nj) == (3, 3):
                    reward = 0

                value = reward + gamma * V[ni, nj]

                action_values.append(value)

            # Select best action
            policy[i, j] = np.argmax(action_values)

            if old_action != policy[i, j]:
                policy_stable = False

    # Stop when policy does not change
    if policy_stable:
        break


# -------------------------
# Display Results
# -------------------------

print("Optimal Value Function:")
for row in V:
    print(["{:.2f}".format(x) for x in row])

print("\nOptimal Policy:")

for i in range(rows):
    row = []

    for j in range(cols):

        if (i, j) in terminal_states:
            row.append("G")
        else:
            row.append(action_symbols[policy[i, j]])

    print(row)
```
