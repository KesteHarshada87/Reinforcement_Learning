```python
import numpy as np

# Grid World
rows, cols = 4, 4

# Terminal states
terminal_states = [(0, 0), (3, 3)]

# Actions: Up, Down, Left, Right
actions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
action_symbols = ['↑', '↓', '←', '→']

# Discount factor
gamma = 0.9
theta = 0.001


# Get next state
def get_next_state(state, action):
    i, j = state
    di, dj = actions[action]

    ni = i + di
    nj = j + dj

    # Stay in the same cell if movement is outside the grid
    if ni < 0 or ni >= rows or nj < 0 or nj >= cols:
        ni, nj = i, j

    return ni, nj


# =========================================================
# VALUE ITERATION
# =========================================================

V = np.zeros((rows, cols))

while True:
    delta = 0

    for i in range(rows):
        for j in range(cols):

            if (i, j) in terminal_states:
                continue

            old_value = V[i, j]

            action_values = []

            for action in range(4):

                ni, nj = get_next_state((i, j), action)

                reward = -1

                # Goal state
                if (ni, nj) == (3, 3):
                    reward = 0

                value = reward + gamma * V[ni, nj]

                action_values.append(value)

            # Choose maximum action value
            V[i, j] = max(action_values)

            delta = max(delta, abs(old_value - V[i, j]))

    if delta < theta:
        break


# =========================================================
# EXTRACT OPTIMAL POLICY
# =========================================================

optimal_policy = np.zeros((rows, cols), dtype=int)

for i in range(rows):
    for j in range(cols):

        if (i, j) in terminal_states:
            continue

        action_values = []

        for action in range(4):

            ni, nj = get_next_state((i, j), action)

            reward = -1

            if (ni, nj) == (3, 3):
                reward = 0

            value = reward + gamma * V[ni, nj]

            action_values.append(value)

        optimal_policy[i, j] = np.argmax(action_values)


# =========================================================
# DISPLAY RESULTS
# =========================================================

print("VALUE ITERATION")
print("\nOptimal Value Function:")

for row in V:
    print(["{:.2f}".format(x) for x in row])


print("\nOptimal Policy:")

for i in range(rows):

    row = []

    for j in range(cols):

        if (i, j) in terminal_states:
            row.append("G")
        else:
            row.append(action_symbols[optimal_policy[i, j]])

    print(row)


# =========================================================
# COMPARISON
# =========================================================

print("\nCOMPARISON WITH POLICY ITERATION")

print("----------------------------------------")
print("Method             Result")
print("----------------------------------------")
print("Policy Iteration   Optimal Policy")
print("Value Iteration    Optimal Policy")
print("----------------------------------------")

print("\nBoth methods produce the same optimal policy")
print("for this Grid World problem.")
```
