```python
import numpy as np

# Grid World size
rows, cols = 4, 4

# Value function initialized to 0
V = np.zeros((rows, cols))

# Terminal states
terminal_states = [(0, 0), (3, 3)]

# Actions: Up, Down, Left, Right
actions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

# Discount factor
gamma = 0.9

# Policy: equal probability for all actions
policy_prob = 0.25

# Iterative Policy Evaluation
theta = 0.001

while True:
    delta = 0

    for i in range(rows):
        for j in range(cols):

            # Skip terminal states
            if (i, j) in terminal_states:
                continue

            old_value = V[i, j]
            new_value = 0

            # Evaluate all possible actions
            for action in actions:
                ni = i + action[0]
                nj = j + action[1]

                # If action goes outside grid, stay in same state
                if ni < 0 or ni >= rows or nj < 0 or nj >= cols:
                    ni, nj = i, j

                # Reward for every step
                reward = -1

                # Terminal state reward
                if (ni, nj) == (3, 3):
                    reward = 0

                # Bellman equation
                new_value += policy_prob * (
                    reward + gamma * V[ni, nj]
                )

            V[i, j] = new_value

            delta = max(delta, abs(old_value - new_value))

    # Stop when values become stable
    if delta < theta:
        break

# Display final value function
print("Final Value Function:")

for row in V:
    print(["{:.2f}".format(value) for value in row])
```
